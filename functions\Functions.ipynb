CHECK WHETHER THE GIVEN NUMBER IS PALINDROME OR NOT USING FUNCTION

```python
def palindrome(a):
    original=a
    rev=0
    while(a!=0):
      num=a%10
      rev=rev*10+num
      a=a//10
    if(rev==original):
      return True
    else:
      return False
print(palindrome(1441))
```

```python
def reverseStr(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
s=input("Enter a string: ")
print(reverseStr(s))
```

TYPES OF ARGUMENTS


```python
#DEFAULT ARGUMENTS
def employe(id,salary=50000):
  print("Employe id is",id,"and salary is",salary)
employe(108,75000)
```

```python
#KEYWORD ARGUMENTS AND POSITIONAL ARGUMENTS
def employe(id,salary):
  print("Employe id is",id,"and salary is",salary)
employe(salary=75000,id=108)
```

```python
#VARIABLE LENGTYH ARGUMENTS
def employers(*num):
  print(num)
  print(sum(num))
s=employers(108,109,110)
```

```python
#FUNCTION THAT RETURNS MORE TAN ONE VALUE
def multiple_values(*num):
  print(num)
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
a,b,d=multiple_values(1,2,3,4,5)
print(f"Sum is {a},Count is {b} and Average is {d}")
```
