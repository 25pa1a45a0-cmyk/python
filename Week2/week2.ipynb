```python
num = int(input("Enter number:"))
match num:
  case 1:
    print("Day is Monday")
  case 2:
    print("Day is Tuesday")
  case 3:
    print("Day is Wednesday")
  case 4:
    print("Day is Thursday")
  case 5:
    print("Day is Friday")
  case 6:
    print("Day is Saturday")
  case 7:
    print("Day is Sunday")
  case _:
    print("No Day")
```
