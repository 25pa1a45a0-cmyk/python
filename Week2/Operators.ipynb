OPERATORS IN PYTHON


ARITHMETIC OPERATORS

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))
print(f"sum of {a} and {b} is {a+b}")
print(f"The difference of {a} and {b} is:{a-b}")
print(f"The product of {a} and {b} is {a*b}")
print(f"The quotient of {a} and {b} is {a/b}")
print(f"The remainder of {a} and {b} is {a%b}")
print(f"The floor divion of {a} and {b} is {a//b}")
print(f"The exponent of {a} and {b} is {a**b}")
print(-1.8/3)
print(2.0%2)
```

COMPARISON OPERATORS

```python
a=10
b=15
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))
a+=b
print(a)
a-=b
print(a)
a*=b
print(a)
a/=b
print(a)
a%=b
print(a)
a//=b
print(a)
a**=b
print(a)

```

IDENTITY OPERATORS

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))
print(a is b)
print(a is not b)
"""sequence type,numeric type,bool,frozenset
 acts as mutable datatype means  value with same data has same corresponding address"""
print(id(a))
print(id(b))
k=[1,2,3]
m=[1,2,3]
print(k is m)
""""List,dict,set, acts as immutable datatypes in
 which they are dynamically allocated in real time their address will differ(change)"""
 #Identity operators are used to compare the addresses rather than their values.
 #The id() function is used to retrive the address of the value where it is stored
print(k is not m)
print(id(k))
print(id(m))
```

TERNARY OPERATOR

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))
result = f"{a} is greater than b" if a>b else f"{b} is greater than {a}"
print(result)
```

BITWISE OPERATORS  
1,BITWISE AND(&)  
2,BITWISE OR(|)  
3,BITWISE XOR(^)  
4,LEFT SHIFT(<<)  
5,RIGHT SHIFT(>>)

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))
print(a&b)
print(a|b)
print(a^b)
print(a<<b)
print(a>>b)
```

LOGICAL OPERATORS  
1,AND(&&)  
2,OR(|)  
3,NOT(!)

```python
a=int(input("Enter a number:"))
b=int(input("Enter another number:"))
c=int(input("Enter another number:"))
print(a>b and a>c)
print(a>b or a>c)
print(not(a>b and a>c))

```

DIFFERENT WAYS OF PRINTING IN CONSOLE USING PRINT FUNCTION

```python
name="Sri"
num=100
w=45.6
print("My name is",name,"My num is",num,"My weight is",w)
print("My name is "+name+" My num is "+str(num)+" My weight is "+str(w))
print("My name is {} My num is {} My weight is {}".format(name,num,w))
print(f"My name is {name} My num is {num} My weight is {w}")
#\n to print in newline
print("My name is {}\nMy num is {} \nMy weight is {}".format(name,num,w))
```

```python
#The below code executed in two new lines because of interpreter which executes code line by line
print("My name is Sri")
print("MY num is 100")
```

```python
#To execute code in single line with two or more print functions, we use end=" ",this in print statement
print("My name is Sri and",end=" ")
print("My num is 100")
```
