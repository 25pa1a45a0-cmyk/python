Odd numbers

```python
n=int(input("Enter number:"))                   #odd numbers
for i in range(1,n,2):
  print(i,end=" ")
```

Multiplication Table

```python
n=int(input("Enter number:"))                  #multiplication table
for i in range(1,11):
  print(n,"*",i ,"=" ,i*n)
```

The given number is prime or not

```python
n = int(input("Enter a number:"))              #the given number is prime or not
for i in range(2,n):
  if(n%i==0):
    print("It is not a prime")
    break
else:
  print("It is a prime")
```

Prime Numbers between two numbers

```python
k = int(input("Enter a number:"))                 #prime numbers between two numbers
l = int(input("Enter ending number:"))
for i in range(k,l+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i,end=" ")


```

Sum of digits of a given number


```python
n=int(input("Enter a number:"))
sum=0
while(n!=0):
  digit=n%10
  sum=sum+digit
  n=n//10
print(sum)
```
